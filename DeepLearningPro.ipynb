```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

headers = {
    "User-Agent": "Mozilla/5.0"
}

category_links = [
    "https://www.calories.info/food/potato-products",
    "https://www.calories.info/food/vegetables-legumes",
    "https://www.calories.info/food/mushrooms",
    "https://www.calories.info/food/salad",
    "https://www.calories.info/food/fruit",
    "https://www.calories.info/food/milk-dairy-products",
    "https://www.calories.info/food/yogurt",
    "https://www.calories.info/food/cheese",
    "https://www.calories.info/food/cream-cheese",
    "https://www.calories.info/food/meat",
    "https://www.calories.info/food/beef-veal",
    "https://www.calories.info/food/pork",
    "https://www.calories.info/food/poultry-chicken-turkey",
    "https://www.calories.info/food/ham-sausage",
    "https://www.calories.info/food/rice-products",
    "https://www.calories.info/food/bread-rolls-pastries",
    "https://www.calories.info/food/pasta-noodles",
    "https://www.calories.info/food/flour-grains-baking-ingredients",
    "https://www.calories.info/food/fish-seafood",
    "https://www.calories.info/food/meals-dishes",
    "https://www.calories.info/food/fast-food-burgers",
    "https://www.calories.info/food/pizza",
    "https://www.calories.info/food/sushi",
    "https://www.calories.info/food/sweets-chocolate-cookies-candy",
    "https://www.calories.info/food/ice-cream",
    "https://www.calories.info/food/cakes-pies",
    "https://www.calories.info/food/desserts-pudding",
    "https://www.calories.info/food/nuts-seeds",
    "https://www.calories.info/food/oils-fats",
    "https://www.calories.info/food/tofu-vegan-products",
    "https://www.calories.info/food/supplements-protein-powder",
    "https://www.calories.info/food/sauces-gravy-dressing-spreads",
    "https://www.calories.info/food/chips-popcorn-snacks",
    "https://www.calories.info/food/herbs-spices-tea",
]

data = []

for category_url in category_links:
    print(f"\nScraping: {category_url}")

    category_name = category_url.split("/")[-1]

    res = requests.get(category_url, headers=headers)
    soup = BeautifulSoup(res.text, "html.parser")

    rows = soup.find_all("tr")

    print(f"Rows: {len(rows)}")

    for row in rows:
        name_tag = row.find("a")
        cols = row.find_all("td")


        if name_tag and len(cols) >= 3:
            name = name_tag.text.strip()
            serving = cols[1].text.strip()
            calories = cols[2].text.strip()

            print(name, serving, calories)

            data.append([name, serving, calories, category_name])

    time.sleep(1)


df = pd.DataFrame(data, columns=["Food Name", "Serving", "Calories", "Category"])


df.to_csv("food_data3.csv", index=False)

print("\n🔥 DONE!")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

headers = {
    "User-Agent": "Mozilla/5.0"
}

category_urls = {
    "pizza": "https://tasty.co/search?q=pizza&sort=popular",
    "burger": "https://tasty.co/search?q=burger&sort=popular",
    "chicken": "https://tasty.co/search?q=chicken&sort=popular",
    "rice": "https://tasty.co/search?q=rice&sort=popular",
    "apple": "https://tasty.co/search?q=apple&sort=popular",
    "coffee": "https://tasty.co/search?q=coffee&sort=popular",
    "sushi": "https://tasty.co/search?q=sushi&sort=popular",
}
category_urls.update({
    # chocolate
    "chocolate": "https://tasty.co/search?q=chocolate&sort=popular",
    "cocoa": "https://tasty.co/search?q=cocoa&sort=popular",
    "chocolate cake": "https://tasty.co/search?q=chocolate+cake&sort=popular",

    # beans
    "beans": "https://tasty.co/search?q=beans&sort=popular",
    "black beans": "https://tasty.co/search?q=black+beans&sort=popular",
    "kidney beans": "https://tasty.co/search?q=kidney+beans&sort=popular",
    "lentils": "https://tasty.co/search?q=lentils&sort=popular",

    # extras (مهمة للdataset بتاعك)
    "vanilla": "https://tasty.co/search?q=vanilla&sort=popular",
    "honey": "https://tasty.co/search?q=honey&sort=popular",
    "cinnamon": "https://tasty.co/search?q=cinnamon&sort=popular"
})

all_data = []

def get_soup(url):
    try:
        res = requests.get(url, headers=headers, timeout=10)
        if res.status_code != 200:
            return None
        return BeautifulSoup(res.text, "html.parser")
    except:
        return None


for category, url in category_urls.items():
    print(f"\nScraping: {category}")

    soup = get_soup(url)
    if not soup:
        continue

    links = soup.find_all("a", href=True)

    recipes = set()

    for link in links:
        href = link["href"]

        if "/recipe/" in href:
            name = link.get_text(strip=True)

            if name:
                recipes.add((name, "https://tasty.co" + href))

    print(f"Found {len(recipes)} recipes")

    for name, link in list(recipes)[:30]:  # limit for stability
        soup2 = get_soup(link)
        if not soup2:
            continue

        ingredients = soup2.find_all("li")

        ing_list = []
        for i in ingredients:
            text = i.get_text(strip=True)
            if any(word in text.lower() for word in ["cup", "tsp", "tbsp", "salt", "oil", "flour", "egg", "milk"]):
                ing_list.append(text)

        if ing_list:
            all_data.append([name, category, ", ".join(ing_list)])

        time.sleep(0.5)

df = pd.DataFrame(all_data, columns=["Recipe Name", "Category", "Ingredients"])
df.drop_duplicates(inplace=True)

df.to_csv("tasty_clean_dataset.csv", index=False)

print(f"\nDONE ✔ Total recipes: {len(df)}")
```

**Encoder**

```python
encoder = models.Sequential([
    layers.Input(shape=(224,224,3)),

    layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    layers.MaxPooling2D((2,2))
])
```

**Decoder**

```python
decoder = models.Sequential([
    layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    layers.UpSampling2D((2,2)),

    layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    layers.UpSampling2D((2,2)),

    layers.Conv2D(3, (3,3), activation='sigmoid', padding='same')
])
```

**Combine**

```python
autoencoder = models.Sequential([
    encoder,
    decoder
])
```

```python
import os
from pathlib import Path
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import train_test_split


# ── 8 classes ────────────────────────────────────────────────────────────────
# Dynamically get class names from the dataset directory
DATASET_BASE_PATH = "dataset/dataset" # Define the base path here
# Get class names from actual subdirectories
CLASS_NAMES = sorted([d for d in os.listdir(DATASET_BASE_PATH) if os.path.isdir(os.path.join(DATASET_BASE_PATH, d))])
NUM_CLASSES   = len(CLASS_NAMES)
CLASS_TO_IDX  = {name: i for i, name in enumerate(CLASS_NAMES)}
IDX_TO_CLASS  = {i: name for name, i in CLASS_TO_IDX.items()}

IMAGE_SIZE    = 224

# ── ImageNet mean/std (نفس ما اتدرب ViT عليه) ───────────────────────────────
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

# ── Transforms ───────────────────────────────────────────────────────────────
# أكتر augmentation عشان الداتا صغيرة (55 صورة/class بس)
train_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE + 32, IMAGE_SIZE + 32)),
    transforms.RandomCrop(IMAGE_SIZE),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.1),
    transforms.ColorJitter(brightness=0.3, contrast=0.3,
                           saturation=0.3, hue=0.05),
    transforms.RandomRotation(degrees=15),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])

val_transforms = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
])


# ── Dataset Class ─────────────────────────────────────────────────────────────
class FoodDataset(Dataset):

    VALID_EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp"}

    def __init__(self, samples: list, transform=None):
        self.samples   = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, label

    @classmethod
    def from_directory(cls, root_dir: str, transform=None):
        """بيبني الـ dataset من فولدر root مباشرة"""
        root    = Path(root_dir)
        samples = []

        # Iterate over CLASS_NAMES which is now dynamically generated
        for class_name in CLASS_NAMES: # Use CLASS_NAMES directly
            idx = CLASS_TO_IDX[class_name] # Get index from map
            class_folder = root / class_name
            if not class_folder.exists():
                print(f"[تحذير] مش موجود: {class_folder}")
                continue
            found = 0
            for img_file in class_folder.iterdir():
                if img_file.suffix.lower() in cls.VALID_EXTENSIONS:
                    samples.append((str(img_file), idx))
                    found += 1
            print(f"  {class_name}: {found} صورة")

        print(f"[Dataset] الإجمالي: {len(samples)} صورة\n")
        return cls(samples, transform)


# ── Helper: split + build dataloaders ────────────────────────────────────────
def build_dataloaders(
    root_dir: str,
    batch_size: int = 16,
    val_split: float = 0.15,
    test_split: float = 0.10,
    num_workers: int = 2,
    seed: int = 42,
):
    """
    بيرجع dict فيه: train / val / test DataLoaders

    FIX 1: الـ bug الأصلي — train_val_labels اتعمل معاها كـ X مش كـ stratify
    FIX 2: batch_size=16 عشان الداتا صغيرة (55 صورة/class)
    FIX 3: زيادة الـ augmentation تعويضاً عن قلة البيانات
    """
    temp_ds     = FoodDataset.from_directory(root_dir, transform=None)
    all_samples = temp_ds.samples
    all_labels  = [s[1] for s in all_samples]

    if len(all_samples) == 0:
        raise ValueError("مفيش صور اتلقت! تأكد من مسار الـ dataset")

    # Step 1: Split off test set
    train_val_samples, test_samples, _, _ = train_test_split(
        all_samples,
        all_labels,
        test_size=test_split,
        stratify=all_labels,
        random_state=seed,
    )

    # Step 2: Split train / val
    # FIX: استخرج الـ labels من train_val_samples الجديد مش المتغير القديم
    tv_labels = [s[1] for s in train_val_samples]
    relative_val = val_split / (1 - test_split)

    train_samples, val_samples = train_test_split(
        train_val_samples,
        test_size=relative_val,
        stratify=tv_labels,
        random_state=seed,
    )

    print(f"[Split] Train: {len(train_samples)} | "
          f"Val: {len(val_samples)} | "
          f"Test: {len(test_samples)}")
    print(f"        (~{len(train_samples)//NUM_CLASSES} / "
          f"{len(val_samples)//NUM_CLASSES} / "
          f"{len(test_samples)//NUM_CLASSES} per class)\n")

    train_ds = FoodDataset(train_samples, transform=train_transforms)
    val_ds   = FoodDataset(val_samples,   transform=val_transforms)
    test_ds  = FoodDataset(test_samples,  transform=val_transforms)

    loaders = {
        "train": DataLoader(train_ds, batch_size=batch_size, shuffle=True,
                            num_workers=num_workers, pin_memory=True),
        "val":   DataLoader(val_ds,   batch_size=batch_size, shuffle=False,
                            num_workers=num_workers, pin_memory=True),
        "test":  DataLoader(test_ds,  batch_size=batch_size, shuffle=False,
                            num_workers=num_workers, pin_memory=True),
    }
    return loaders


# ── Quick sanity check ────────────────────────────────────────────────────────
if __name__ == "__main__":
    import sys
    # Explicitly set the root directory to the correct dataset path.
    root = DATASET_BASE_PATH # Use the defined base path

    print("=" * 50)
    print(" Dataset Sanity Check")
    print("=" * 50)

    loaders = build_dataloaders(root, batch_size=8, num_workers=0)

    imgs, labels = next(iter(loaders["train"]))
    print(f"Batch shape : {imgs.shape}")
    print(f"Labels      : {labels.tolist()}")
    print(f"Classes     : {[IDX_TO_CLASS[l.item()] for l in labels]}")
    print(f"Pixel range : [{imgs.min():.2f}, {imgs.max():.2f}]")
    print("\nAll good! ✓")
```

```python
import pickle
import numpy as np
import pandas as pd
from pathlib import Path
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# ══════════════════════════════════════════════════════════
# Config
# ══════════════════════════════════════════════════════════
MODEL_NAME   = "all-MiniLM-L6-v2"   # سريع + خفيف + دقيق
EMBEDDINGS_PATH = "food_embeddings.pkl"
CSV_PATH        = "final_matched_dataset (1).csv"
```

```python
def load_and_clean_csv(csv_path: str) -> pd.DataFrame:
    df = pd.read_csv(csv_path)

    # تنظيف الأسماء
    df["name"] = df["name"].str.strip().str.lower()

    # حذف الصفوف اللي مفيهاش calories
    df = df.dropna(subset=["calories"])
    df = df[df["calories"] > 0]

    # تعبئة الـ nulls بـ 0 في الـ macros
    for col in ["protein", "fat", "carb"]:
        df[col] = df[col].fillna(0)

    # نعمل عمود نص موحد للـ embedding: name + ingredients + category
    df["text_for_embedding"] = (
        df["name"].fillna("") + " " +
        df["ingredients"].fillna("") + " " +
        df["category"].fillna("")
    )

    # نشيل الـ duplicates بناءً على الاسم وناخد الـ average للـ calories
    df = df.groupby("name", as_index=False).agg({
        "text_for_embedding": "first",
        "ingredients":        "first",
        "category":           "first",
        "portion":            "first",
        "calories":           "mean",
        "protein":            "mean",
        "fat":                "mean",
        "carb":               "mean",
    })

    df["calories"] = df["calories"].round(1)
    df["protein"]  = df["protein"].round(1)
    df["fat"]      = df["fat"].round(1)
    df["carb"]     = df["carb"].round(1)

    print(f"[CSV] {len(df):,} صنف أكل بعد التنظيف")
    return df.reset_index(drop=True)

```

```python
# 2. بناء الـ Embeddings وحفظها
# ══════════════════════════════════════════════════════════
def build_and_save_embeddings(df: pd.DataFrame, model: SentenceTransformer):
    """
    بيعمل embed لكل الأسماء ويحفظها — بيتعمل مرة واحدة بس
    """
    print(f"[Embedding] بيعمل embed لـ {len(df):,} صنف ...")
    texts      = df["text_for_embedding"].tolist()
    embeddings = model.encode(
        texts,
        batch_size=256,
        show_progress_bar=True,
        normalize_embeddings=True,   # مهم عشان cosine similarity يبقى أسرع
    )

    with open(EMBEDDINGS_PATH, "wb") as f:
        pickle.dump({"embeddings": embeddings, "df": df}, f)

    print(f"[Embedding] اتحفظ في {EMBEDDINGS_PATH}")
    return embeddings


def load_embeddings():
    """بيحمّل الـ embeddings المحفوظة"""
    with open(EMBEDDINGS_PATH, "rb") as f:
        data = pickle.load(f)
    print(f"[Embedding] اتحمّل من {EMBEDDINGS_PATH}")
    return data["embeddings"], data["df"]

```

```python
# 3. الـ Search Function
# ══════════════════════════════════════════════════════════
def search_food(
    query: str,
    model: SentenceTransformer,
    embeddings: np.ndarray,
    df: pd.DataFrame,
    top_k: int = 3,
) -> list[dict]:
    """
    بيرجع أقرب top_k نتايج للـ query

    Returns:
        list of dicts, كل dict فيه معلومات الأكل
    """
    query_clean = query.strip().lower()

    # Embed الـ query
    query_vec = model.encode(
        [query_clean],
        normalize_embeddings=True,
    )

    # Cosine similarity مع كل الـ embeddings
    scores = cosine_similarity(query_vec, embeddings)[0]

    # Top K indices
    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []
    for idx in top_indices:
        row = df.iloc[idx]
        results.append({
            "name":        row["name"],
            "category":    row["category"],
            "portion":     row["portion"],
            "calories":    row["calories"],
            "protein":     row["protein"],
            "fat":         row["fat"],
            "carb":        row["carb"],
            "ingredients": row["ingredients"],
            "score":       round(float(scores[idx]), 3),
        })

    return results

```

```python
# 4. عرض النتايج
# ══════════════════════════════════════════════════════════
def format_result(result: dict, rank: int = 1) -> str:
    return (
        f"\n{'='*45}\n"
        f"  #{rank}  {result['name'].title()}\n"
        f"{'='*45}\n"
        f"  الفئة      : {result['category']}\n"
        f"  الحصة      : {result['portion']}\n"
        f"  السعرات    : {result['calories']} kcal\n"
        f"  بروتين     : {result['protein']} g\n"
        f"  دهون       : {result['fat']} g\n"
        f"  كارب       : {result['carb']} g\n"
        f"  تطابق      : {result['score']*100:.1f}%\n"
    )


# ══════════════════════════════════════════════════════════
# 5. Chatbot Loop
# ══════════════════════════════════════════════════════════
def run_chatbot(model, embeddings, df):
    print("\n" + "="*45)
    print("  🍕 Food Calories Chatbot")
    print("  اكتب اسم أي أكل — اكتب 'exit' للخروج")
    print("="*45 + "\n")

    while True:
        query = input("أنت: ").strip()

        if not query:
            continue
        if query.lower() in ("exit", "quit", "خروج"):
            print("مع السلامة!")
            break

        results = search_food(query, model, embeddings, df, top_k=3)

        if not results or results[0]["score"] < 0.3:
            print("  ⚠️  مش لاقي نتيجة قريبة. جرب اسم تاني.\n")
            continue

        for i, r in enumerate(results, 1):
            print(format_result(r, rank=i))

```

```python
# 6. Main
# ══════════════════════════════════════════════════════════
if __name__ == "__main__":
    # تحميل الـ model
    print(f"[Model] بيحمّل {MODEL_NAME} ...")
    model = SentenceTransformer(MODEL_NAME)

    # تحميل أو بناء الـ embeddings
    if Path(EMBEDDINGS_PATH).exists():
        embeddings, df = load_embeddings()
    else:
        df         = load_and_clean_csv(CSV_PATH)
        embeddings = build_and_save_embeddings(df, model)

    # شغّل الـ chatbot
    run_chatbot(model, embeddings, df)
```

```python
import numpy as np
import pickle
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

CSV_PATH = "final_matched_dataset (1).csv"

# ── Test cases ──────────────────────────────────────────
TEST_CASES = [
    {"query": "pizza",              "expected": "pizza"},
    {"query": "burger",             "expected": "burger"},
    {"query": "grilled chicken",    "expected": "chicken"},
    {"query": "green salad",        "expected": "salad"},
    {"query": "french fries",       "expected": "fries"},
    {"query": "scrambled eggs",     "expected": "egg"},
    {"query": "white rice",         "expected": "rice"},
    {"query": "banana",             "expected": "banana"},
    {"query": "coffee",             "expected": "coffee"},
    {"query": "pasta",              "expected": "pasta"},
    {"query": "sushi",              "expected": "sushi"},
    {"query": "ice cream",          "expected": "ice cream"},
    {"query": "koshary",            "expected": "koshari"},
    {"query": "falafel",            "expected": "falafel"},
    {"query": "shawarma",           "expected": "shawarma"},
    {"query": "hummus",             "expected": "hummus"},
    {"query": "kebab",              "expected": "kebab"},
    {"query": "chips",              "expected": "fries"},
    {"query": "fried rice",         "expected": "rice"},
    {"query": "chocolate cake",     "expected": "chocolate"},
]

MODELS_TO_TEST = [
    "all-MiniLM-L6-v2",                          # الـ model القديم
    "all-mpnet-base-v2",                          # أقوى
    "paraphrase-multilingual-MiniLM-L12-v2",      # يفهم عربي
]

# ── Load & clean CSV ────────────────────────────────────
import pandas as pd
df = pd.read_csv(CSV_PATH)
df["name"] = df["name"].str.strip().str.lower()
df = df.dropna(subset=["calories"])
df = df[df["calories"] > 0]
for col in ["protein","fat","carb"]:
    df[col] = df[col].fillna(0)
df["text_for_embedding"] = (
    df["name"].fillna("") + " " +
    df["ingredients"].fillna("") + " " +
    df["category"].fillna("")
)
df = df.groupby("name", as_index=False).agg({
    "text_for_embedding": "first",
    "calories": "mean", "protein": "mean",
    "fat": "mean", "carb": "mean",
    "category": "first", "portion": "first",
})
df = df.reset_index(drop=True)
texts = df["text_for_embedding"].tolist()

# ── Evaluate each model ─────────────────────────────────
print("=" * 60)
print("  MODEL COMPARISON")
print("=" * 60)

best_model_name = None
best_top1       = 0

for model_name in MODELS_TO_TEST:
    print(f"\n⏳ بيجرّب: {model_name}")
    model      = SentenceTransformer(model_name)
    embeddings = model.encode(texts, batch_size=256,
                              normalize_embeddings=True,
                              show_progress_bar=True)

    top1, top3 = 0, 0
    for tc in TEST_CASES:
        vec    = model.encode([tc["query"].lower()], normalize_embeddings=True)
        scores = cosine_similarity(vec, embeddings)[0]
        idx    = np.argsort(scores)[::-1][:3]
        names  = [df.iloc[i]["name"] for i in idx]
        exp    = tc["expected"].lower()
        if exp in names[0] or names[0] in exp: top1 += 1
        if any(exp in n or n in exp for n in names): top3 += 1

    t1 = top1 / len(TEST_CASES) * 100
    t3 = top3 / len(TEST_CASES) * 100
    print(f"  ✅ Top-1: {t1:.1f}%  |  Top-3: {t3:.1f}%")

    if t1 > best_top1:
        best_top1       = t1
        best_model_name = model_name
        # احفظ الـ embeddings بتاع أحسن model
        import pickle
        with open("food_embeddings_best.pkl", "wb") as f:
            pickle.dump({"embeddings": embeddings, "df": df,
                         "model_name": model_name}, f)

print(f"\n{'='*60}")
print(f"  🏆 أحسن model: {best_model_name}")
print(f"  🏆 Top-1 Accuracy: {best_top1:.1f}%")
print(f"  💾 اتحفظ في food_embeddings_best.pkl")
print(f"{'='*60}")
```
